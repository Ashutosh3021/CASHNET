```python
path = kagglehub.dataset_download("ealaxi/paysim1")
print("Path to dataset files:", path)

drive_path = '/content/drive/MyDrive/CASHNET/paysim1'
for item in os.listdir(path):
    s = os.path.join(path, item)
    d = os.path.join(drive_path, item)
    if os.path.isdir(s):
        shutil.copytree(s, d, dirs_exist_ok=True)
    else:
        shutil.copy2(s, d)

print(f"Files copied to: {drive_path}")
```

```python
zip_path = '/content/drive/MyDrive/CASHNET/ieee-fraud-detection/ieee-fraud-detection.zip'
extract_path = '/content/drive/MyDrive/CASHNET/ieee-fraud-detection'

if os.path.exists(zip_path):
    print(f"Extracting {zip_path}...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_path)
    
    print("Extraction complete. Deleting zip file...")
    os.remove(zip_path)
    print("Zip file deleted.")
else:
    print(f"Zip file not found at {zip_path}")
```
